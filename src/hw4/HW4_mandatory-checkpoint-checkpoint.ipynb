**Ключевые метрики.**

```python
def precision_at_k(ranked_doc_ids, relevant_set, k):
    top_k = ranked_doc_ids[:k]
    if k == 0: return 0.0
    return sum(1 for d in top_k if d in relevant_set) / k

def average_precision_at_k(ranked_doc_ids, relevant_set, k):
    if not relevant_set:
        return 0.0
    hits, score = 0, 0.0
    for i, d in enumerate(ranked_doc_ids[:k], start=1):
        if d in relevant_set:
            hits += 1
            score += hits / i
    denom = min(k, len(relevant_set))
    return score / denom if denom > 0 else 0.0

def dcg(gains):
    return sum((2**g - 1) / np.log2(i + 2) for i, g in enumerate(gains))

def ndcg_at_k(ranked_doc_ids, qrels_for_query, k):
    gains = [qrels_for_query.get(d, 0) for d in ranked_doc_ids[:k]]
    ideal_gains = sorted(qrels_for_query.values(), reverse=True)[:k]
    idcg = dcg(ideal_gains)
    return dcg(gains) / idcg if idcg > 0 else 0.0

def evaluate_on_metric(run, qrels):
    p1, p10, p20, ap20, ndcg2 = [], [], [], [], []
    for qid, ranked in run.items():
        rel = qrels.get(qid, {})
        rel_set = {d for d, g in rel.items() if g > 0}
        if not rel_set:
            continue
        p1.append(precision_at_k(ranked, rel_set, 1))
        p10.append(precision_at_k(ranked, rel_set, 10))
        p20.append(precision_at_k(ranked, rel_set, 20))
        ap20.append(average_precision_at_k(ranked, rel_set, 20))
        ndcg2.append(ndcg_at_k(ranked, rel, 2))
    
    return {
        "P@1":    float(np.mean(p1)),
        "P@10":   float(np.mean(p10)),
        "P@20":   float(np.mean(p20)),
        "MAP@20": float(np.mean(ap20)),
        "nDCG@2": float(np.mean(ndcg2)),
        "num_queries": len(p1),
    }
```

```python
def load_mirage(seed=SEED):
    """
      corpus: {doc_id: text}
      queries: {qid: text}
      qrels:  {qid: {doc_id: 1}}  (binary; 'support'==1 marks relevant)
    Corpus = union of doc chunks across all query pools (deduplicated by text hash).
    """
    data = load_dataset("nlpai-lab/mirage", split="train")
    data = data.shuffle(seed=seed).select(range(len(data)))


    corpus, queries, qrels = {}, {}, {}
    text_to_id = {}

    def doc_id_for(text, hint):
        if text in text_to_id:
            return text_to_id[text]
        did = f"d{len(text_to_id)}_{hint[:40].replace(' ', '_')}"
        text_to_id[text] = did
        corpus[did] = text
        return did

    for row in data:
        qid = row["query_id"]
        queries[qid] = row["query"]
        qrels[qid] = {}

        pool = row["doc_pool"]
        names = pool["doc_name"]
        chunks = pool["doc_chunk"]
        supports = pool.get("support", [0] * len(chunks))

        for name, chunk, sup in zip(names, chunks, supports):
            did = doc_id_for(chunk, name)
            if int(sup) > 0:
                qrels[qid][did] = 1
                
    print(f"MIRAGE. Всего queries: {len(queries)}, размер корпуса документов: {len(corpus)}")
    return corpus, queries, qrels

mirage_corpus, mirage_queries, mirage_qrels = load_mirage()
```

```python
def evaluate(run, qrels):
    
    p1, p10, p20, ap20, ndcg2 = [], [], [], [], []
    
    for qid, ranked in run.items():
        rel = qrels.get(qid, {})
        rel_set = {d for d, g in rel.items() if g > 0}
        if not rel_set:
            continue
        p1.append(precision_at_k(ranked, rel_set, 1))
        p10.append(precision_at_k(ranked, rel_set, 10))
        p20.append(precision_at_k(ranked, rel_set, 20))
        ap20.append(average_precision_at_k(ranked, rel_set, 20))
        ndcg2.append(ndcg_at_k(ranked, rel, 2))
    return {
        "P@1": float(np.mean(p1)),
        "P@10": float(np.mean(p10)),
        "P@20": float(np.mean(p20)),
        "MAP@20": float(np.mean(ap20)),
        "nDCG@2": float(np.mean(ndcg2)),
        "num_queries": len(p1),
    }
```

**To же самое для WikIR.**

```python
def load_wikir():
    ds = ir_datasets.load("wikir/en1k/test")

    corpus = {d.doc_id: (d.text if hasattr(d, "text") else d.body) for d in ds.docs_iter()}
    queries = {q.query_id: q.text for q in ds.queries_iter()}
    qrels = defaultdict(dict)
    for qrel in ds.qrels_iter():
        qrels[qrel.query_id][qrel.doc_id] = int(qrel.relevance)

    qrels = {q: {d: g for d, g in r.items()} for q, r in qrels.items()
             if any(g > 0 for g in r.values())}
    queries = {q: t for q, t in queries.items() if q in qrels}

    print(f"WIKIR. Всего queries: {len(queries)}, размер корпуса документов: {len(corpus)}\n")
    return corpus, queries, dict(qrels)

wikir_corpus, wikir_queries, wikir_qrels = load_wikir()
```

У нас есть wikir_corpus, wikir_queries, wikir_qrels, mirage_corpus, mirage_queries, mirage_qrels (чтобы не забыть)

```python
def encode_texts(model, ids, texts, batch_size=64, cache_key=None):
    if cache_key:
        p = CACHE_DIR / f"{cache_key}.pt"
        if p.exists():
            obj = torch.load(p, map_location=DEVICE)
            return obj["ids"], obj["emb"].to(DEVICE)

    emb = model.encode(
        texts,
        batch_size=batch_size,
        convert_to_tensor=True,
        show_progress_bar=True,
        normalize_embeddings=True,
        device=DEVICE,
    )
    if cache_key:
        torch.save({"ids": ids, "emb": emb.cpu()}, CACHE_DIR / f"{cache_key}.pt")
    return ids, emb


def run_retrieval(model_name, corpus, queries, top_k=20, dataset_tag="ds"):
    print(f"\nRUN {model_name} on {dataset_tag}")
    model = SentenceTransformer(model_name, device=DEVICE)

    short = model_name.split("/")[-1]
    corp_ids = list(corpus.keys())
    corp_texts = [corpus[d] for d in corp_ids]
    q_ids = list(queries.keys())
    q_texts = [queries[q] for q in q_ids]
    
    _, corp_emb = encode_texts(model, corp_ids, corp_texts,
                               cache_key=f"{dataset_tag}_{short}_corpus"
                              )
    _, q_emb = encode_texts(model, q_ids, q_texts,
                            cache_key=f"{dataset_tag}_{short}_queries"
                           )

    hits = util.semantic_search(q_emb, corp_emb, top_k=top_k,
                                score_function=util.cos_sim)

    run = {}
    for qid, hitlist in zip(q_ids, hits):
        run[qid] = [corp_ids[h["corpus_id"]] for h in hitlist]
    
    del model, corp_emb, q_emb
    gc.collect()
    if DEVICE == "cuda":
        torch.cuda.empty_cache()
    return run
```

```python
DATASETS = {
    "MIRAGE": (mirage_corpus, mirage_queries, mirage_qrels),
    "WIKIR":  (wikir_corpus,  wikir_queries,  wikir_qrels),
}
```

```python
RERANK_MODELS = [
    "cross-encoder/ms-marco-MiniLM-L-6-v2",
    "cross-encoder/ms-marco-MiniLM-L-12-v2",
]
K_VALUES = [20, 50]
```

```python
from sentence_transformers import CrossEncoder
import re
from rank_bm25 import BM25Okapi
from tqdm.notebook import tqdm
```

Бейзлайн

```python
def tokenize(text):
    return re.findall(r"\w+", text.lower())

def build_bm25(corpus):
    
    corp_ids = list(corpus.keys())
    print("start tokenizing")
    
    tokens = [tokenize(corpus[d]) for d in tqdm(corp_ids, desc = "tokenize corpus for simple bm25 run")]
    start_tok = time.time()
    print("finished tokenization. start bm25 initialization")
    bm25 = BM25Okapi(tokens)
    print(f"ended bm25 initialization  at {time.time() - start_tok}")
    return bm25, corp_ids


def bm25_simple(bm25, corp_ids, queries, k):
    print("start simple bm25")
    
    q_ids = list(queries.keys())
    start = time.time()
    
    scores_matrix = np.vstack([
        bm25.get_scores(tokenize(queries[q_idx])) for q_idx in tqdm(q_ids)
    ])
    top_idx = np.argsort(-scores_matrix, axis=1)[:, :k]
    run = {
        q_idx: [(corp_ids[idx], float(scores_matrix[i, idx])) for idx in tqdm(top_idx[i])]
        for i, q_idx in enumerate(q_ids)
    }
    return run, time.time() - start
    
```

```python
def rerank(bm25_run, queries, corpus, model_name, top_k, batch_size=128):
    print("start reranker")
    model = CrossEncoder(model_name, device=DEVICE, max_length=512)
    
    q_ids = list(bm25_run.keys())
    candidates = [bm25_run[q][:top_k] for q in q_ids]
    lengths = [len(c) for c in candidates]

    pairs = [
        (queries[q], corpus[doc_id])
        for q, cands in zip(q_ids, candidates)
        for doc_id, _ in cands
    ]

    start = time.time()
    print("model rerank START")
    all_scores = model.predict(pairs, batch_size=batch_size, show_progress_bar=True)
    print("model rerank FINISH")
    total_time = time.time() - start

    rerun, ptr = {}, 0
    for q, cands, cand_len in tqdm(zip(q_ids, candidates, lengths), desc="reranking "):
        scores = all_scores[ptr: ptr + cand_len]
        order  = np.argsort(-scores)
        rerun[q] = [cands[i][0] for i in order]
        ptr += cand_len
    del model
    gc.collect()
    if DEVICE == "cuda":
        torch.cuda.empty_cache()
    return rerun, total_time, len(pairs)
```

Прогон всех конфигураций

```python
rerank_results = []

for dataset_name, (corpus, queries, qrels) in DATASETS.items():
    print(f"\n{dataset_name}")
    bm25, corp_ids = build_bm25(corpus)
    bm25_run_max, bm25_time = bm25_simple(bm25, corp_ids, queries, k=max(K_VALUES))
    
    for k in tqdm(K_VALUES):
        bm25_run_k = {q: [d for d, _ in v[:k]] for q, v in bm25_run_max.items()}
        bm25_metrics = evaluate_v2(bm25_run_k, qrels)
        rerank_results.append({
            "dataset": ds_name, "model": "BM25", "k": k,
            "bm25_time_s": round(bm25_time, 3), "rerank_time_s": 0.0,
            "total_time_s": round(bm25_time, 3),
            "pairs_scored": 0, "ms_per_pair": 0.0,
            **bm25_metrics,
        })
    
    print(f"  BM25 baseline: {bm25_metrics}  [BM25 time: {bm25_time:.1f}s]")

    for model_name in tqdm(RERANK_MODELS):
        short = model_name.split("/")[-1]
        for k in K_VALUES:
            rerun, rr_time, n_pairs = rerank(bm25_run_max, queries, corpus, model_name, top_k=k)
            metrics = evaluate(rerun, qrels)
            per_pair_ms = 1000 * rr_time / max(n_pairs, 1)
            rerank_results.append({
                "dataset": ds_name, "model": short, "k": k,
                "bm25_time_s": bm25_time, "rerank_time_s": rr_time,
                "total_time_s": bm25_time + rr_time,
                "pairs_scored": n_pairs,
                "ms_per_pair": per_pair_ms,
                **metrics,
            })
            print(f"  {short} k={k}: {metrics}  "
                  f"[rerank {rr_time:.1f}s, {per_pair_ms:.2f} ms/pair]")

rerank_df = pd.DataFrame(rerank_results)
```
